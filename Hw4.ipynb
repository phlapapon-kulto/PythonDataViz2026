### เขียน function ตัดเกรด โดยที่ input เป็นคะแนน(0-100) , output เป็นเกรด (F-A) วนลูปตัดเกรด input = [1,50,65,90,101,-5,49]

```python
scores = ['1','50','65','90','101','-5','49']
```

```python
def grades(score):
  grade_list = []
  for score_val in scores: # Renamed loop variable to avoid shadowing the function parameter
    current_score = int(score_val)
    if current_score >= 0 and current_score <= 50:
      print("Grade F")
    elif current_score > 50 and current_score <= 55: # Corrected indentation and range for 50
      print("Grade D")
    elif current_score > 55 and current_score <= 60:
      print("Grade D+")
    elif current_score > 60 and current_score <= 65:
      print("Grade C")
    elif current_score > 65 and current_score <= 70:
      print("Grade C+")
    elif current_score > 70 and current_score <= 75:
      print("Grade B")
    elif current_score > 75 and current_score <= 80:
      print("Grade B+")
    elif current_score > 80 and current_score <= 100:
      print("Grade A")
    else:
      print("Error")
      grade_list.append(current_score)
  return grade_list
```

```python
print("ข้อมูลคะแนน:",scores)
print("ข้อมูลเกรด:")
print(grades(scores))
```
