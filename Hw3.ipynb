### ให้ สร้าง list ของ เวลา ```['12:30:15','13:41:07',....]``` แล้ววนลูปเรียกฟังก์ชั่นคำนวณเวลาเป็น วินาที



````python
def time_second(list_time):
  second_list = []
  for time in list_time:
      hours, minutes, second = map(int, time.split(':'))
      total_second = hours * 3600 + minutes * 60 + second
      second_list.append(total_second)
  return second_list
````

````python
print("ข้อมูลเวลา:",list_time)
print("ข้อมูลเวลา(วินาที):",time_second(list_time))
````
