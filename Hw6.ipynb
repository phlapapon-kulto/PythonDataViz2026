```python
import os
path = '/content/drive/My Drive/PythonDataViz2026'
file_path = os.path.join(path,'confirmed-cases.csv')
print(file_path)
```

```python
data_covid = pd.read_csv(file_path)
data_covid.head()
```

```python
data_covid.shape
```

```python
data = data_covid[['sex','age','province_of_onset']]
data
```

```python
female_data = data[data['sex'] == 'หญิง']
female_numeric_age = pd.to_numeric(female_data['age'], errors='coerce')
print(female_numeric_age.head())
```

```python
sum = 0
N = 0
for a in female_numeric_age:
    if a > 0:
        sum += a
        N += 1
print(f'อายุเฉลี่ยของผู้ป่วยหญิง:{sum/N}')
```

```python
female_kk = female_data[female_data['province_of_onset'] == 'ขอนแก่น']
female_kk_numeric = pd.to_numeric(female_kk['age'], errors='coerce')
print(female_kk_numeric.head())
```

```python
sum = 0
N = 0
for a in female_kk_numeric :
    if a > 0:
        sum += a
        N += 1

print(f'อายุเฉลี่ยของผู้ป่วยหญิงในจังหวัดขอนแก่น {sum/N}')
```

```python
male_data = data[data['sex'] == 'ชาย']
male_numeric_age = pd.to_numeric(male_data['age'], errors='coerce')
print(male_numeric_age.head())
```

```python
sum = 0
N = 0
for a in male_numeric_age:
    if a > 0:
        sum += a
        N += 1
print(f'อายุเฉลี่ยของผู้ป่วยชาย:{sum/N}')
```

```python
male_kk = male_data[male_data['province_of_onset'] == 'ขอนแก่น']
male_kk_numeric = pd.to_numeric(male_kk['age'], errors='coerce')
print(male_kk_numeric.head())
```

```python
sum = 0
N = 0
for a in male_kk_numeric :
    if a > 0:
        sum += a
        N += 1

print(f'อายุเฉลี่ยของผู้ป่วยชายในจังหวัดขอนแก่น {sum/N}')
```
